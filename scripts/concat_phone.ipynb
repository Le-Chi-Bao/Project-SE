```python
import pandas as pd
import os

# Đường dẫn từ scripts/ đến data/
df1 = pd.read_csv('../data/reprocess/cleaned_phone_CellphoneS.csv')
df2 = pd.read_csv('../data/reprocess/cleaned_phone_thegioididong.csv')
df3 = pd.read_csv('../data/reprocess/cleaned_phone_tiki.csv')

# Thêm cột nguồn
df1['data_source'] = 'CellphoneS'
df2['data_source'] = 'TheGioiDiDong'
df3['data_source'] = 'Tiki'

# Concat tất cả features
combined_df = pd.concat([df1, df2, df3], ignore_index=True, join='outer')

print("✅ Concat thành công!")
print(f"Shape: {combined_df.shape}")
print(f"Tổng số sản phẩm: {len(combined_df)}")
print(f"Tổng số features: {len(combined_df.columns)}")
```

```python
# Kiểm tra chi tiết từng cột
print("\n🔍 PHÂN TÍCH CÁC FEATURES TỪ 3 NGUỒN:")

for col in combined_df.columns:
    missing_count = combined_df[col].isnull().sum()
    missing_percent = (missing_count / len(combined_df)) * 100
    print(f"  {col}: {missing_count} missing ({missing_percent:.1f}%)")
```

```python
print("\n📊 FEATURES ĐẶC THÙ THEO NGUỒN:")

sources = ['CellphoneS', 'TheGioiDiDong', 'Tiki']
for source in sources:
    source_df = combined_df[combined_df['data_source'] == source]
    non_missing_cols = source_df.columns[source_df.notna().any()].tolist()
    print(f"  {source}: {len(non_missing_cols)} features")
```

```python
combined_df.to_csv('../final_data_phone.csv', index=False, encoding='utf-8-sig')
print("💾 Đã xuất file: final_data_phone.csv")

# Hoặc nếu muốn đầy đủ hơn
output_filename = "../final_data_phone.csv"
combined_df.to_csv(output_filename, index=False, encoding='utf-8-sig')
print(f"🎉 Đã lưu file: {output_filename}")
print(f"📁 Số dòng: {len(combined_df)}, Số cột: {len(combined_df.columns)}")
```
