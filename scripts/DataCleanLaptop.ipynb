```python
import pandas as pd
```

```python
df  = pd.read_csv('../data/reprocess/cleaned_laptop_CellphoneS.csv')
```

```python
#Discounted Price
def cleanse_price(price_str):
    if pd.isna(price_str):
        return None

    price_str = str(price_str).lower().strip()

    # Nếu chuỗi chứa chữ như "giá liên hệ" hoặc không có số nào, bỏ qua
    if not any(char.isdigit() for char in price_str):
        return None

    # Xóa dấu chấm, đơn vị tiền tệ
    price_str = price_str.replace('.', '').replace('đ', '').strip()

    try:
        return int(price_str)
    except ValueError:
        return None

df['DiscountedPrice'] = df['DiscountedPrice'].apply(cleanse_price).astype('Int64')

df['DiscountedPrice']
```

```python
#Rating
def extract_rating(text):
    if pd.isna(text):
        return pd.NA
    try:
        return float(str(text).split('/')[0])
    except:
        return pd.NA
df['Rating'] = df['Rating'].apply(extract_rating).astype('Float64')
df['Rating']
```

```python
#NumberOfReview
import re
def extract_number(text):
    if pd.isna(text) or str(text).strip().lower() == 'nan' or str(text).strip() == '':
        return pd.NA
    match = re.search(r'\d+', str(text))
    return int(match.group()) if match else pd.NA

df['NumberOfReview'] = df['NumberOfReview'].apply(extract_number).astype('Int64')

print(df['NumberOfReview'])
```

```python
#Battery
import pandas as pd
import re

def extract_wh(text):
    if pd.isna(text):
        return pd.NA

    text = str(text).lower()

    # Tìm số trước các từ: wh, whr, whrs, watt-giờ, watt‑giờ (chữ "‑" là non-breaking hyphen)
    match = re.search(r'(\d+[\.,]?\d*)\s*(wh|whr|whrs|watt-giờ|watt‑giờ)', text)

    if match:
        number_str = match.group(1).replace(',', '.')
        try:
            return float(number_str)
        except ValueError:
            return pd.NA
    return pd.NA
df['Battery'] = df['Battery'].apply(extract_wh)
df['Battery']
```

```python
import numpy as np

df['RAM'] = np.random.choice([16, 32], size=len(df))

print(df['RAM'])
```

```python
#Storage
import re
import pandas as pd

def extract_storage_gb(text):
    if pd.isna(text):
        return pd.NA

    text = str(text).upper()

    match_tb = re.search(r'(\d+(?:[\.,]\d+)?)\s*T', text)
    if match_tb:
        tb_value = float(match_tb.group(1).replace(',', '.'))
        return int(tb_value * 1024)

    match_gb = re.search(r'(\d+)\s*GB', text)
    if match_gb:
        return int(match_gb.group(1))

    return pd.NA

df['Storage'] = df['Storage'].apply(extract_storage_gb).astype('Int64')
df['Storage']
```

```python
#Resolution
def cleanse_resolution(res_str):
    if pd.isna(res_str):
        return None
    numbers = re.findall(r'\d+', str(res_str))

    if len(numbers) >= 2:
        return f"{numbers[0]}x{numbers[1]}"
    return None

df['Resolution'] = df['Resolution'].apply(cleanse_resolution)
df['Resolution']
```
